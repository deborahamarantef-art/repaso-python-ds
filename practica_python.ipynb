Manipulación de datos básicos en Python: variables, bucles, condicionales, diccionarios y funciones.

## Tarea 1: Definición de variables

```python
# Variables de un producto con distintos tipos de datos
nombre_producto = "Auriculares inalámbricos"  # str
precio = 89.90                                # float
stock = 25                                    # int
tiene_descuento = True                        # bool

# Resumen del producto usando f-strings
print(f"Producto: {nombre_producto} | Precio: ${precio:.2f} | Stock: {stock} unidades | Descuento disponible: {'Sí' if tiene_descuento else 'No'}")
```

## Tarea 2: Lógica de negocio (bucle + condicional)

```python
# Lista de precios a clasificar
precios = [45.5, 120.0, 99.99, 250.0, 100, 15.75]

# Recorremos cada precio y lo clasificamos según el umbral de 100
for p in precios:
    if p > 100:
        print(f"${p:.2f} -> Caro")
    else:
        print(f"${p:.2f} -> Económico")
```

## Tarea 3: Estructura de datos compleja (diccionario)

```python
# Diccionario de stock: producto -> cantidad
almacen = {
    "Teclado": 40,
    "Mouse": 75,
    "Monitor": 12,
}

# Agregamos un nuevo producto
almacen["Webcam"] = 30

# Consultamos el stock de un producto específico
producto_consulta = "Monitor"
print(f"Stock de {producto_consulta}: {almacen[producto_consulta]} unidades")
print("Almacén actualizado:", almacen)
```

## Tarea 4: Modularización con funciones

```python
def resumen_estadistico(numeros):
    """Recibe una lista de números y devuelve un diccionario con suma, promedio y cantidad."""
    cantidad = len(numeros)
    suma = sum(numeros)
    promedio = suma / cantidad if cantidad > 0 else 0  # evita división por cero con listas vacías
    return {
        "suma": suma,
        "promedio": promedio,
        "cantidad": cantidad,
    }

# Llamada con una lista de ejemplo
resultado = resumen_estadistico([10, 20, 30])
print(resultado)

# También podemos aplicarla a la lista de precios de la Tarea 2
print(resumen_estadistico(precios))
```
